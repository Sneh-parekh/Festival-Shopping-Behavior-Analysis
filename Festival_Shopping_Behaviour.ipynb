```python
bins = [0, 25, 40, 60, 100]
labels = ['Young Adult', 'Adult', 'Mid-Aged', 'Senior']

df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)
```

```python
df[['Age', 'Age_Group']].head(10)
```

```python
Frequency_Mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['Purchase_Frequency_Days'] = df['Frequency_of_Purchases'].map(Frequency_Mapping)
```

```python
df = df.drop(columns=['Frequency_of_Purchases'])
```

```python
new_order = [
    # Customer info
    'Customer_ID', 'Age', 'Age_Group', 'Gender', 'Location', 'Subscription_Status',

    # Customer behavior
    'Previous_Purchases', 'Purchase_Frequency_Days',

    # Product info
    'Item_Purchased', 'Category', 'Size', 'Color', 'Season',

    # Transaction
    'Quantity', 'Purchase_Amount_(USD)', 'Discount_Applied',

    # Operations
    'Payment_Method', 'Shipping_Type',

    # Experience
    'Review_Rating'
]

df = df[new_order]
```

## **Visualizations**
---

### **1. Customer Demographics**

```python
# ------------------------
# 1.  Age Distribution
# ------------------------

plt.figure()
sns.histplot(df['Age'], bins=15, kde=True, color='skyblue')
plt.title("Customer Age Distribution", fontsize=14, fontweight='bold')
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# ------------------------
# Gender Split
# ------------------------

plt.figure()
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=['lightgreen','lightcoral'])
plt.title("Gender Split of Customers", fontsize=14, fontweight='bold')
plt.ylabel('')
plt.show()
```

```python
# ------------------------
# Subscription by Age Group
# ------------------------

plt.figure()
pd.crosstab(df['Age_Group'], df['Subscription_Status']).plot(kind='bar', stacked=True)
plt.title("Subscription Status by Age Group", fontsize=14, fontweight='bold')
plt.ylabel("Number of Customers")
plt.xlabel("Age Group")
plt.show()
```

### **2. Product Popularity**

```python
# ------------------------
# Top 10 Purchased Items
# ------------------------


top_items = df.groupby('Item_Purchased')['Quantity'].sum().sort_values(ascending=False).head(10)
plt.figure()
top_items.plot(kind='barh', color='orange')
plt.title("Top 10 Purchased Items by Quantity", fontsize=14, fontweight='bold')
plt.xlabel("Quantity Sold")
plt.ylabel("Items")
plt.gca().invert_yaxis()
plt.show()
```

```python
# ------------------------
# Sales by Category
# ------------------------

cat_sales = df.groupby('Category')['Purchase_Amount_(USD)'].sum().sort_values(ascending=False)
plt.figure()
cat_sales.plot(kind='bar', color='purple')
plt.title("Total Sales by Category", fontsize=14, fontweight='bold')
plt.ylabel("Revenue (USD)")
plt.show()
```

### **3. Revenue & Discounts**

```python
# ------------------------
# Plot: Average Purchase Amount vs Discount by Category
# ------------------------

plt.figure(figsize=(10,5))
sns.lineplot(data=df, x='Discount_Applied', y='Purchase_Amount_(USD)',
             hue='Category', estimator='mean', marker='o', palette='Set2')
plt.title("Average Purchase Amount vs Discount by Category", fontsize=14, fontweight='bold')
plt.xlabel("Discount Applied (%)")
plt.ylabel("Average Purchase Amount (USD)")
plt.legend(title='Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()

```

### **4. Customer Behavior**

```python
# ------------------------
# Purchase Frequency Distribution
# ------------------------

plt.figure()
sns.histplot(df['Purchase_Frequency_Days'], bins=20, color='teal')
plt.title("Purchase Frequency Distribution (Days)", fontsize=14, fontweight='bold')
plt.xlabel("Days Between Purchases")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# ------------------------
# Top 10 Customers by Revenue
# ------------------------

top_customers = df.groupby('Customer_ID')['Purchase_Amount_(USD)'].sum().sort_values(ascending=False).head(10)
plt.figure()
top_customers.plot(kind='barh', color='gold')
plt.title("Top 10 Customers by Revenue", fontsize=14, fontweight='bold')
plt.xlabel("Revenue (USD)")
plt.ylabel("Customer ID")
plt.gca().invert_yaxis()
plt.show()
```

### **5. Customer Experience**

```python
# ------------------------
# Review Ratings Distribution
# ------------------------

plt.figure()
sns.countplot(data=df, x='Review_Rating', palette='pastel')
plt.title("Distribution of Customer Review Ratings", fontsize=14, fontweight='bold')
plt.xlabel("Review Rating")
plt.ylabel("Number of Reviews")
plt.show()
```

```python
# ------------------------
# Average Rating by Category
# ------------------------

avg_rating = df.groupby('Category')['Review_Rating'].mean().sort_values()
plt.figure()
avg_rating.plot(kind='bar', color='skyblue')
plt.title("Average Customer Rating by Category", fontsize=14, fontweight='bold')
plt.ylabel("Average Rating")
plt.show()
```
